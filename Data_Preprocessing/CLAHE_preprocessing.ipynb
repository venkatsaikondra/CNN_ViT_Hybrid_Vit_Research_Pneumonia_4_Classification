```python
import cv2
import os

def apply_clahe_to_directory(input_root, output_root, clip_limit=2.0, tile_grid_size=(8, 8)):
    """
    Applies CLAHE preprocessing to images in input_root and saves to output_root.
    Maintains the internal folder structure.
    """
    # Initialize CLAHE
    clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=tile_grid_size)
    valid_exts = ('.jpg', '.jpeg', '.png', '.bmp', '.tiff')

    if not os.path.exists(input_root):
        print(f"Error: Source folder not found at {input_root}")
        return

    print(f"Processing images from: {input_root}")

    for root, _, files in os.walk(input_root):
        for file in files:
            if file.lower().endswith(valid_exts):
                # Setup Paths
                input_path = os.path.join(root, file)

                # Create corresponding output subdirectory
                rel_path = os.path.relpath(root, input_root)
                dest_dir = os.path.join(output_root, rel_path)
                os.makedirs(dest_dir, exist_ok=True)

                output_path = os.path.join(dest_dir, file)

                # Process Image
                img = cv2.imread(input_path)
                if img is None:
                    continue

                # CLAHE requires a single channel (Grayscale)
                gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
                processed_img = clahe.apply(gray)

                # Save
                cv2.imwrite(output_path, processed_img)

    print(f"Success! Processed data saved to: {output_root}")

# --- CONFIGURATION FOR JUPYTER ---
# Get the parent directory of 'Data_Preprocessing'
project_root = os.path.dirname(os.getcwd())
# --- DIAGNOSTIC CONFIGURATION ---
project_root = os.path.dirname(os.getcwd())

# 1. Manually double-check this spelling against your sidebar!
# Change this if your images are actually in the other folder
input_dir = os.path.join(project_root, "Vit_Augumented_Data")
output_dir = os.path.join(project_root, "Vit_Augumented_Data_CLAHE")

# 2. Check if the folder is actually empty before starting
if os.path.exists(input_dir):
    all_files = []
    for root, dirs, files in os.walk(input_dir):
        for f in files:
            all_files.append(f)

    print(f"Total files found in source: {len(all_files)}")
    if len(all_files) > 0:
        print(f"Example file found: {all_files[0]}")
        # Run the function only if files exist
        apply_clahe_to_directory(input_dir, output_dir)
    else:
        print("ALERT: Source folder is empty. Check your path or spelling!")
else:
    print(f"ALERT: The path {input_dir} does not exist at all.")
```
