```python
import splitfolders
import os

# --- PATH CONFIGURATION ---
# Get project root (moving up from Data_Augumentation folder)
project_root = os.path.dirname(os.getcwd())

# Source: The folder containing your 4 processed classes
input_folder = os.path.join(project_root, "ViT_Augumented_Data_CLAHE")

# Destination: The new split folder
output_folder = os.path.join(project_root, "Train_Test_SPLit_ViT_Data_CLAHE")

# --- THE SPLIT ---
# ratio=(train, validation, test)
splitfolders.ratio(
    input_folder,
    output=output_folder,
    seed=42,
    ratio=(.8, .1, .1),
    group_prefix=None,
    move=False # Set to True if you want to move files instead of copying
)

print(f"Split complete! Data saved to: {output_folder}")
```
